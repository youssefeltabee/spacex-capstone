```python
# Task 2: Mark success (green) and failed (red) launches for each site
# Load full launch dataset from earlier EDA
launch_data_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"
launch_df = pd.read_csv(launch_data_url)

# Fix column name mismatch (launch_df uses 'LaunchSite' without space, geo df uses 'Launch Site')
launch_df = launch_df.rename(columns={'LaunchSite': 'Launch Site'})

# Normalize site name formatting (replace hyphens with spaces to match geo dataset)
launch_df['Launch Site'] = launch_df['Launch Site'].str.replace('-', ' ', regex=False)
spacex_df['Launch Site'] = spacex_df['Launch Site'].str.replace('-', ' ', regex=False)

# Merge to get coordinates for each launch
launch_geo_df = pd.merge(
    launch_df,
    spacex_df[['Launch Site', 'Lat', 'Long']],
    on='Launch Site',
    how='left'
).dropna(subset=['Lat', 'Long'])

# Create marker cluster for launches
marker_cluster = MarkerCluster().add_to(site_map)

# Add success (green) and failed (red) markers
for _, row in launch_geo_df.iterrows():
    color = 'green' if row['Class'] == 1 else 'red'
    folium.CircleMarker(
        location=[row['Lat'], row['Long']],
        radius=5,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.7,
        popup=f"Site: {row['Launch Site']}<br>Outcome: {'Success' if row['Class']==1 else 'Failure'}"
    ).add_to(marker_cluster)

# Add legend
legend_html = '''
<div style="position:fixed; bottom:50px; left:50px; width:150px; height:90px;
background-color:white; border:2px solid grey; z-index:9999; font-size:14px; padding:10px;">
<b>Launch Outcome</b><br>
<span style="color:green;">●</span> Success<br>
<span style="color:red;">●</span> Failure
</div>
'''
site_map.get_root().html.add_child(folium.Element(legend_html))
site_map
```

```python
# Task 3: Calculate distances between launch sites and mark them on the map
from math import radians, sin, cos, sqrt, atan2

def haversine(lat1, lon1, lat2, lon2):
    """Calculate distance between two coordinates in kilometers"""
    R = 6371  # Earth radius in km
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = sin(dlat/2)**2 + cos(lat1) * cos(lat2) * sin(dlon/2)**2
    c = 2 * atan2(sqrt(a), sqrt(1-a))
    return R * c

# Calculate pairwise distances between unique launch sites
print("Pairwise distances between launch sites (km):")
sites = launch_sites_df[['Launch Site', 'Lat', 'Long']].values.tolist()
for i in range(len(sites)):
    for j in range(i+1, len(sites)):
        site1, lat1, lon1 = sites[i]
        site2, lat2, lon2 = sites[j]
        dist = haversine(lat1, lon1, lat2, lon2)
        print(f"{site1} ↔ {site2}: {dist:.1f} km")

# Add distance lines between sites on the map
for i in range(len(sites)):
    for j in range(i+1, len(sites)):
        site1, lat1, lon1 = sites[i]
        site2, lat2, lon2 = sites[j]
        folium.PolyLine(
            locations=[[lat1, lon1], [lat2, lon2]],
            color='blue',
            weight=2,
            opacity=0.5,
            popup=f"Distance: {haversine(lat1, lon1, lat2, lon2):.1f} km"
        ).add_to(site_map)

site_map
```

### Analysis Findings
1. **Proximity to Equator**: 
   - CCAFS LC-40, CCAFS SLC-40, and KSC LC-39A are at ~28.5°N (low latitude, favorable for launches due to higher rotational speed)
   - VAFB SLC-4E is at ~34.6°N (higher latitude, not near the equator)
   - Most sites are at relatively low latitudes compared to higher northern latitudes, optimizing launch efficiency.

2. **Proximity to Coast**:
   - All 4 launch sites are directly located on the coast (Florida sites on the Atlantic coast, VAFB on the Pacific coast)
   - This is critical for safety (debris falls into water) and for ocean landings of Falcon 9 first stages.

3. **Distances Between Sites**:
   - Florida-based sites (CCAFS LC-40, CCAFS SLC-40, KSC LC-39A) are all within ~7 km of each other
   - VAFB SLC-4E is ~3,700 km from the Florida launch sites
